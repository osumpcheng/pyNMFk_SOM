```python
import seaborn as sns

df_alpha = pd.DataFrame(
    np.transpose(np.array([np.append(df_sites_qc_top.sum(axis=1).values,df_sites_qc_btm.sum(axis=1).values),
    np.append(np.repeat('surface soil', df_sites_qc_top.shape[0]), np.repeat('subsoil', df_sites_qc_btm.shape[0]))])),
    columns = ['alpha_diversity', 'layer'])

df_alpha['alpha_diversity'] = pd.to_numeric(df_alpha['alpha_diversity'])

#sns.set(rc={'figure.figsize':(5,8)})
sns.set(font_scale=1.5)
ax = sns.boxplot(data=df_alpha, x='layer', y='alpha_diversity')
ax.set_xlabel("Layer", fontsize=18)
ax.set_ylabel("\u03B1 diversity", fontsize=18)
#ax.set_xticklabels(ax.get_xticklabels(), fontsize=14)
#ax.set_yticklabels(ax.get_yticklabels(), fontsize=14)
sns.despine(left=True)
```

```python
import scipy.stats as stats

f_statistic, p_value = stats.f_oneway(df_alpha[df_alpha['layer']=='surface soil']['alpha_diversity'], df_alpha[df_alpha['layer']=='subsoil']['alpha_diversity'])

print("F-Statistic: ", f_statistic)
print("P-Value: ", p_value)
```

```python
def mixup_data(x1, x2, y1, y2):

    lam   = 0.5 #Sample from beta distribution
    x_lam = lam * x1 + (1.0 - lam) * x2 #Mixup inputs vectors
    y_lam = lam * y1 + (1.0 - lam) * y2 #Mixup output vectors

    return x_lam, y_lam

X_mix = [[None] * 7]
y_mix = []
for i in np.arange(len(y_train)):
    for j in np.arange(i+1,len(y_train)):
        x_new, y_new = mixup_data(X_train[i,:], X_train[j,:], y_train[i], y_train[j])
        X_mix = np.append(X_mix, [x_new], axis=0)
        y_mix = np.append(y_mix, y_new)
    print(i)
    print(X_mix.shape)
    print(y_mix.shape)
```
